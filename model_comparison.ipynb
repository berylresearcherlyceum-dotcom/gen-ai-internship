## Task 2: One Question, Four Answers
## How each model generates new data??

* **Autoencoder:** It doesn’t truly generate new data, it only reconstructs what it has seen.

* **VAE:** It samples from a learned probability space and decodes it into new data.

* **GAN:** It converts random noise into realistic data using a trained generator.

* **Diffusion:** It starts from noise and gradually removes it to create data.

* **LLM:** It generates data by predicting the next word one token at a time.



## PART B — Mapping to LLMs (Critical)
## Task 3: How LLMs Are Generative Models

## What is being predicted in an LLM?

An LLM predicts the next token based on all previous tokens. At every step, it answers: “What token is most likely to come next?”

## Why token prediction = generation

Once a token is predicted, it is added back into the input and used to predict the next one. Repeating this step many times builds a full sentence or paragraph, which is why simple prediction becomes full text generation.

## How probability is involved

The model does not choose a token randomly or deterministically. It outputs a probability distribution over all possible tokens, and one token is sampled from this distribution, allowing both correctness and creativity.

## Task 4: Prompt → Probability → Output

```python
import random

def sample_token(token_probs):
    tokens = list(token_probs.keys())
    probs = list(token_probs.values())
    return random.choices(tokens, probs)[0]

# Example prompt
prompt = "The capital of France is"
print("Prompt:", prompt)

# What the "model" predicts
next_token_probs = {
    "Paris": 0.85,
    "Lyon": 0.05,
    "Berlin": 0.01,
    ".": 0.09
}

# Generate output multiple times
for i in range(5):
    next_token = sample_token(next_token_probs)
    print(f"Run {i+1} →", prompt, next_token)
```

## PART C — Hands-on
## Task 5: Sampling Demonstration

```python
import numpy as np

# Define a simple probability distribution
tokens = ["A", "B", "C"]
probabilities = [0.6, 0.3, 0.1]  # Probabilities sum to 1

# Sample 10 times
samples = np.random.choice(tokens, size=10, p=probabilities)
print("Samples:", samples)

# Show approximate frequency
for token in tokens:
    freq = np.sum(samples == token)
    print(f"{token}: {freq} times")
```

## Task 6: Temperature Experiment

```python
import numpy as np

# Tokens and base probabilities
tokens = ["A", "B", "C"]
probs = np.array([0.6, 0.3, 0.1])  # Sum = 1

def sample_with_temperature(probs, temperature=1.0, n_samples=10):
    # Apply temperature: scale logits before softmax
    logits = np.log(probs + 1e-8) / temperature  # avoid log(0)
    exp_logits = np.exp(logits)
    adjusted_probs = exp_logits / np.sum(exp_logits)
    return np.random.choice(tokens, size=n_samples, p=adjusted_probs)

# Low temperature (deterministic-like)
low_temp_samples = sample_with_temperature(probs, temperature=0.1)
print("Low temperature samples:", low_temp_samples)

# High temperature (diverse)
high_temp_samples = sample_with_temperature(probs, temperature=2.0)
print("High temperature samples:", high_temp_samples)
```
