## PART D — MINI PROJECT (System Thinking)
## Mini Project: Choosing the Right Generative Model

| Application            | Best Model        | Why Others Don’t Fit                                           |
| ---------------------- | ----------------- | -------------------------------------------------------------- |
| Text Chatbot           | LLM (GPT)         | GANs/VAEs → text is blurry; Diffusion → for images             |
| Image Generation       | Diffusion         | GANs → mode collapse; VAE → blurry; LLM → can’t make pixels    |
| Synthetic Tabular Data | VAE / Tabular GAN | Diffusion → slow; LLM → not structured                         |
| Anomaly Detection      | VAE / Autoencoder | GANs → unstable; Diffusion → heavy                             |
| Creative Writing       | LLM (high temp)   | GANs → not fluent; VAE → fuzzy text; Diffusion → mainly images |

## Task 7: Design a GenAI System 

1. Input:

User provides a prompt: text, image description, or tabular specification.

2. Model Choice:

Text → LLM
Image → Diffusion Model
Tabular → VAE / Tabular GAN

3. Sampling Strategy:

Text: Temperature + top-k/top-p sampling for diversity
Image: Stepwise reverse diffusion
Tabular: Sample from latent space

4. Output:

Generated content: text, image, or table
