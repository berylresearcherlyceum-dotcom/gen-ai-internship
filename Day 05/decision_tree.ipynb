## Task 7: Decision Tree using sklearn

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt

# Sample dataset
X = [[2,5], [3,6], [4,5], [5,7], [6,6], [7,8], [8,7], [3,4], [6,5], [4,6]]
y = [0,0,0,1,1,1,1,0,1,0]

# ----------------------
# Default Decision Tree
# ----------------------
default_tree = DecisionTreeClassifier(random_state=42)
default_tree.fit(X, y)

# Plot default tree
plt.figure(figsize=(12,6))
plot_tree(default_tree, feature_names=['Study Hours','Sleep Hours'], class_names=['Fail','Pass'], filled=True)
plt.title("Default Decision Tree")
plt.show()

# ----------------------
# Restricted Decision Tree (max_depth=2)
# ----------------------
restricted_tree = DecisionTreeClassifier(max_depth=2, random_state=42)
restricted_tree.fit(X, y)

# Plot restricted tree
plt.figure(figsize=(12,6))
plot_tree(restricted_tree, feature_names=['Study Hours','Sleep Hours'], class_names=['Fail','Pass'], filled=True)
plt.title("Restricted Decision Tree (max_depth=2)")
plt.show()

# Predictions
print("Default tree predictions:", default_tree.predict(X))
print("Restricted tree predictions:", restricted_tree.predict(X))
```

## Task 8: Overfitting Experiment

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

# Sample dataset
X = [[2,5], [3,6], [4,5], [5,7], [6,6], [7,8], [8,7], [3,4], [6,5], [4,6]]
y = [0,0,0,1,1,1,1,0,1,0]

# Split into training and testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Train trees with different depths
depths = [1, 2, 5, None]  # None = default (fully grown)
for d in depths:
    tree = DecisionTreeClassifier(max_depth=d, random_state=42)
    tree.fit(X_train, y_train)
    train_acc = tree.score(X_train, y_train)
    test_acc = tree.score(X_test, y_test)
    print(f"max_depth={d}: Train Accuracy={train_acc:.2f}, Test Accuracy={test_acc:.2f}")
```

## Task 9: Tree Visualization

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt

# Sample dataset
X = [[2,5], [3,6], [4,5], [5,7], [6,6], [7,8], [8,7], [3,4], [6,5], [4,6]]
y = [0,0,0,1,1,1,1,0,1,0]

# Train a decision tree
tree = DecisionTreeClassifier(max_depth=3, random_state=42)
tree.fit(X, y)

# Plot the tree
plt.figure(figsize=(12,6))
plot_tree(
    tree, 
    feature_names=['Study Hours','Sleep Hours'], 
    class_names=['Fail','Pass'], 
    filled=True
)
plt.title("Decision Tree Visualization")
plt.show()
```
