## Task 5: Define Routing Rules

```python
# AI Model Router function
def choose_model(feature_type, data_size, need_probabilities):
    """
    Decide which model to use based on input characteristics.
    
    Parameters:
    - feature_type: 'text' or 'numeric'
    - data_size: number of samples
    - need_probabilities: True if probabilities are needed
    
    Returns:
    - Model name as string
    """
    if feature_type == "text":
        return "Naive Bayes"            # Text data → Naive Bayes
    elif feature_type == "numeric" and data_size <= 50:
        return "k-NN"                   # Small numeric dataset → k-NN
    elif need_probabilities:
        return "Logistic Regression"    # Need probabilities → Logistic Regression
    else:
        return "Naive Bayes"            # Default fallback
```

```python
# Test cases
print(choose_model("text", 30, False))       # Expected: Naive Bayes
print(choose_model("numeric", 20, False))    # Expected: k-NN
print(choose_model("numeric", 100, True))    # Expected: Logistic Regression
print(choose_model("text", 10, True))        # Expected: Naive Bayes
```

## Task 6: Implement Router Function

```python
def route_model(input_features):
    """
    Decide which model to use based on input characteristics.

    Parameters:
    - input_features: dict with keys:
        - 'feature_type' : 'text' or 'numeric'
        - 'data_size' : number of samples
        - 'need_probabilities' : True/False

    Returns:
    - chosen_model : string
    """
    feature_type = input_features.get('feature_type')
    data_size = input_features.get('data_size')
    need_probabilities = input_features.get('need_probabilities', False)

    # Apply the routing logic
    if feature_type == "text":
        chosen_model = "Naive Bayes"
        reason = "Text data → Naive Bayes handles textual features well"
    elif feature_type == "numeric" and data_size <= 50:
        chosen_model = "k-NN"
        reason = "Small numeric dataset → k-NN works well on small numeric data"
    elif need_probabilities:
        chosen_model = "Logistic Regression"
        reason = "Probabilities needed → Logistic Regression provides probabilities"
    else:
        chosen_model = "Naive Bayes"
        reason = "Default fallback"

    # Print input characteristics, chosen model, and reason
    print("Input characteristics:", input_features)
    print("Selected model:", chosen_model)
    print("Reason for selection:", reason)
    print("-" * 50)  # separator

    return chosen_model
```

```python
# Example inputs
test_inputs = [
    {'feature_type': 'text', 'data_size': 30, 'need_probabilities': False},
    {'feature_type': 'numeric', 'data_size': 20, 'need_probabilities': False},
    {'feature_type': 'numeric', 'data_size': 100, 'need_probabilities': True},
    {'feature_type': 'text', 'data_size': 10, 'need_probabilities': True}
]

# Test the router
for inp in test_inputs:
    route_model(inp)
```
