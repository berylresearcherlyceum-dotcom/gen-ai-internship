## Task 1: Gradient Sanity Check

```python
import numpy as np

x = 1.5
y = 1
w = 0.8
b = -0.3
eps = 1e-5

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def loss(w):
    z = w * x + b
    y_hat = sigmoid(z)
    return -(y * np.log(y_hat) + (1 - y) * np.log(1 - y_hat))

# Numerical gradient
grad_num = (loss(w + eps) - loss(w - eps)) / (2 * eps)

# Analytical gradient
z = w * x + b
y_hat = sigmoid(z)
grad_analytical = (y_hat - y) * x

print(f"Analytical gradient: {grad_analytical:.8f}")
print(f"Numerical gradient:  {grad_num:.8f}")
print(f"Absolute difference: {abs(grad_analytical - grad_num):.2e}")
```

### Why Gradient Checking Matters in Deep Networks

In deep networks, there are many layers and calculations, so it’s easy to make small mistakes when writing backpropagation code. Even if the program runs, a wrong gradient can stop the model from learning properly. Also, gradient checking compares the calculated gradients with numerical estimates to make sure they are almost the same, so gradient checking matters in deep learning.
