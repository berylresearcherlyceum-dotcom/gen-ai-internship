## Task 9: Prompt Encoding

```python
import numpy as np
import re

# -------- Example prompts --------
prompts = [
    "Write a short story about a dragon.",
    "Explain VAE in simple terms, with examples.",
    "List 5 tips for coding efficiently!"
]

# -------- Function to extract numeric features --------
def encode_prompt(prompt):
    length = len(prompt)                         # total characters
    keywords = len(re.findall(r'\b(a|the|an|is|for|with|5)\b', prompt, re.I))  # simple keyword count
    punctuation = len(re.findall(r'[.,!?]', prompt))
    verbs = len(re.findall(r'\b(write|explain|list)\b', prompt, re.I))        # simple verb count
    return np.array([length, keywords, punctuation, verbs], dtype=float)

# -------- Encode all prompts --------
encoded_prompts = np.array([encode_prompt(p) for p in prompts])

print("Encoded prompt feature vectors:")
print(encoded_prompts)
```

## Task 10: Generate Variations

```python
import numpy as np

# -------- Example: Encoded prompt features from Task 9 --------
encoded_prompts = np.array([
    [32, 1, 1, 1],
    [41, 3, 2, 1],
    [35, 2, 2, 1]
], dtype=float)

input_dim = encoded_prompts.shape[1]
latent_dim = 2

# -------- Dummy encoder (mean & log_var) --------
# Normally learned via VAE training; using simple linear map for demonstration
W_mu = np.random.randn(input_dim, latent_dim) * 0.1
b_mu = np.zeros(latent_dim)

W_logvar = np.random.randn(input_dim, latent_dim) * 0.1
b_logvar = np.zeros(latent_dim)

def encoder(x):
    mu = x @ W_mu + b_mu
    log_var = x @ W_logvar + b_logvar
    return mu, log_var

# -------- Decoder (linear map for demo) --------
W_dec = np.random.randn(latent_dim, input_dim) * 0.1
b_dec = np.zeros(input_dim)

def decoder(z):
    return z @ W_dec + b_dec

# -------- Sample latent vectors from encoder outputs --------
num_variations = 5
for idx, prompt_vec in enumerate(encoded_prompts):
    mu, log_var = encoder(prompt_vec)
    sigma = np.exp(0.5 * log_var)
    
    print(f"\nOriginal prompt vector {idx+1}:", prompt_vec)
    print("Generated variations:")
    for i in range(num_variations):
        z = mu + sigma * np.random.randn(latent_dim)  # Reparameterization trick
        gen_vec = decoder(z)
        print(f"Variation {i+1}:", gen_vec)
```
