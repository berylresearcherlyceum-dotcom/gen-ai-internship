## Task 9: Encode Prompts

```python
import numpy as np

prompts = [
    "Write a short story about a robot.",
    "Explain photosynthesis in simple words.",
    "List 3 benefits of exercise!"
]

def encode_prompt(prompt):
    length = len(prompt)
    keywords = len([w for w in prompt.split() if w.lower() in ["robot", "story", "photosynthesis", "exercise"]])
    punctuation = sum([1 for c in prompt if c in ".,!?"])
    verbs = len([w for w in prompt.split() if w.lower() in ["write", "explain", "list"]])
    return [length, keywords, punctuation, verbs]

X = np.array([encode_prompt(p) for p in prompts], dtype=float)

# Min-max normalization safely
X_min = X.min(axis=0)
X_max = X.max(axis=0)
X_norm = (X - X_min) / np.where((X_max - X_min)==0, 1, X_max - X_min)

print("Encoded & normalized prompts:\n", X_norm)
```

## Task 10: Latent Space Exploration

```python
import numpy as np

# -----------------------------
# Step 0: Tiny dataset
# -----------------------------
# 5 samples, 4 features
X = np.random.rand(5, 4)
print("Original data:\n", X)

# -----------------------------
# Step 1: Define encoder
# -----------------------------
# Encode 4D input into 2D latent space
W_enc = np.random.rand(4, 2)  # weights
b_enc = np.random.rand(2)     # bias

def encoder(X):
    return X @ W_enc + b_enc   # linear encoding

# -----------------------------
# Step 2: Define decoder
# -----------------------------
# Decode 2D latent vector back to 4D
W_dec = np.random.rand(2, 4)  # weights
b_dec = np.random.rand(4)     # bias

def decoder(Z):
    return Z @ W_dec + b_dec   # linear reconstruction

# -----------------------------
# Step 3: Encode
# -----------------------------
latent_vectors = encoder(X)
print("\nLatent vectors:\n", latent_vectors)

# -----------------------------
# Step 4: Add small noise
# -----------------------------
noise = np.random.normal(0, 0.05, latent_vectors.shape)
noisy_latent = latent_vectors + noise

# -----------------------------
# Step 5: Decode
# -----------------------------
original_recon = decoder(latent_vectors)
noisy_recon = decoder(noisy_latent)

print("\nOriginal reconstruction:\n", original_recon)
print("\nNoisy reconstruction:\n", noisy_recon)

# ---
```
